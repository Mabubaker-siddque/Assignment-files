A student is buying a laptop.
Laptop price: Rs. 85,000
Student has: Rs. 100,000
Student gets a discount of Rs. 5,000
A laptop bag costs Rs. 2,500
The store has a promotion code "SAVE10".
Calculate price after discount, final bill, remaining money, check enough money, check promo code, check bill > 80,000, and check truthy of enough-money result.

```python
laptop_price = 85000
student_money = 100000
discount = 5000
bag_price = 2500
promo_code = "SAVE10"

price_after_discount = laptop_price - discount
final_bill = price_after_discount + bag_price
remaining_money = student_money - final_bill
has_enough_money = student_money >= final_bill
promo_valid = "SAVE10" in promo_code
bill_above_80000 = final_bill > 80000

print(price_after_discount)
print(final_bill)
print(remaining_money)
print(has_enough_money)
print(promo_valid)
print(bill_above_80000)
print(bool(has_enough_money))
```

You are creating a student result calculator.
Mathematics: 85, English: 72, Python: 90, Total possible marks per subject: 100
Student name: "Ahmed", Required passing marks per subject: 50
Calculate total, average, percentage, pass checks, name check, average check, and truthy check. Print all results.

```python
maths_marks = 85
english_marks = 72
python_marks = 90
total_per_subject = 100
student_name = "Ahmed"
passing_marks = 50

total_marks = maths_marks + english_marks + python_marks
average_marks = total_marks / 3
percentage = (total_marks / (total_per_subject * 3)) * 100
passed_maths = maths_marks >= passing_marks
passed_all = maths_marks >= passing_marks and english_marks >= passing_marks and python_marks >= passing_marks
name_has_a = "A" in student_name
average_above_80 = average_marks >= 80

print(total_marks)
print(average_marks)
print(percentage)
print(passed_maths)
print(passed_all)
print(name_has_a)
print(average_above_80)
print(bool(percentage))
```
