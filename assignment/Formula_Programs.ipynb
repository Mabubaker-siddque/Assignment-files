## 7) Compound Interest

```python
P = 1000    # principal amount
r = 0.05    # annual interest rate (decimal)
n = 4       # number of times compounded per year
t = 3       # time in years

A = P * (1 + r / n) ** (n * t)
CI = A - P
print("Total Amount:", A)
print("Compound Interest:", CI)
```

## 11) Quadratic Equation Roots

```python
import math

a = 1
b = -3
c = 2

discriminant = b ** 2 - 4 * a * c

if discriminant >= 0:
    root1 = (-b + math.sqrt(discriminant)) / (2 * a)
    root2 = (-b - math.sqrt(discriminant)) / (2 * a)
    print("Root 1:", root1)
    print("Root 2:", root2)
else:
    real_part = -b / (2 * a)
    imag_part = math.sqrt(-discriminant) / (2 * a)
    print(f"Root 1: {real_part} + {imag_part}i")
    print(f"Root 2: {real_part} - {imag_part}i")
```

## 13) Gravitational Force

```python
G = 6.674e-11   # gravitational constant (N*m^2/kg^2)

m1 = 1000   # mass of object 1 in kg
m2 = 500    # mass of object 2 in kg
r = 2       # distance between centers in meters

F = G * (m1 * m2) / (r ** 2)
print("Gravitational Force:", F, "N")
```

## 19) Work Done by a Force

```python
import math

f = 20      # force in Newtons
d = 10      # distance in meters
theta = 30  # angle between force and direction of movement (degrees)

W = f * d * math.cos(math.radians(theta))
print("Work Done:", W, "J")
```

## 20) Heat Transfer

```python
m = 2       # mass in kg
c = 4186    # specific heat capacity in J/(kg*C) -- water
T1 = 25     # initial temperature in Celsius
T2 = 100    # final temperature in Celsius

deltaT = T2 - T1
Q = m * c * deltaT
print("Heat Transfer:", Q, "J")
```
